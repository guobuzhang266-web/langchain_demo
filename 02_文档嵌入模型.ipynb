文档嵌入模型

```python
from dotenv import load_dotenv#读取.env配置文件中的密钥
from langchain.embeddings.base import init_embeddings
import os # 读取环境变量

load_dotenv(override=True)# 读取.env配置文件中的密钥
DASHSCOPE_BASE_URL=os.getenv("DASHSCOPE_BASE_URL")
DASHSCOPE_API_KEY=os.getenv("DASHSCOPE_API_KEY")
embedding_model=init_embeddings(
    model="openai:qwen3.7-text-embedding",
    api_key=DASHSCOPE_API_KEY,
    base_url=DASHSCOPE_BASE_URL,
    check_embedding_ctx_length=False,#关掉 token id 预处理，直接传字符串
    chunk_size=10#适配百炼单次最多 10 条文本的限制

)
```

句子向量化

```python
text1="你好qwen，我是你的朋友"
embed_docs=embedding_model.embed_query(text1)
print(len(embed_docs))
```

文档的向量化

```python
from langchain_community.document_loaders import CSVLoader
from rich import print as rprint
loader = CSVLoader("../../asset/load/02-load.csv", encoding="utf-8")
docs = loader.load_and_split()# 加载并切分数据
# 存放的是每一个chrunk的embedding。
# rprint(docs)
texts = [doc.page_content for doc in docs]
embeded_docs = embedding_model.embed_documents(texts)
print(len(embeded_docs))
for i in range(len(texts)):
    print(f"{texts[i]}:\n{embeded_docs[i][:3]}",end="\n\n")
```
